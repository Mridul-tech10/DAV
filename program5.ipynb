```python
import numpy as np
import pandas as pd
```

```python
data = {
    "students": ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"],
    "marks": [50, 55, 60, 62, 65, 68, 70, 72, 75, 150],
}
df = pd.DataFrame(data)
```

```python
print("og")
print(df)
```

```python
q1=df["marks"].quantile(0.25)
print(q1)
```

```python
q3=df["marks"].quantile(0.75)
print(q3)
```

```python
iqr=q3-q1
print(iqr)
```

```python
l_l=q1-1.5*iqr
u_l=q3+1.5*iqr
print(l_l)
print(u_l)
```

```python
outliers=df[
    (df["marks"]<l_l)|(df["marks"]>u_l)
]
```

```python
print("outliers")
print(outliers)
```

```python
df["marks_treated"]=np.where(
    df["marks"]>u_l,
    u_l,
    np.where(
        df["marks"]<l_l,
        l_l,
        df["marks"]
    )
)
print(df)
```

```python
df_cap=df.copy()
df_cap["marks"]=df_cap["marks"].clip(
    lower=l_l,
    upper=u_l
)
print("\n after cap")
print(df_cap)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
data = {
    "students": ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m", "n", "o", "p", "q", "r", "s", "t"],
    "marks": [50, 52, 54, 55, 56, 58, 60, 61, 62, 63, 64, 65, 66, 68, 70, 71, 72, 73, 75, 200]}
df = pd.DataFrame(data)
df["z_score"] = zscore(df["marks"])
print("z score")
print(df)
```

```python
outliers=df[abs(df["z_score"])>3]
print("outlier z score")
print(outliers)
```

```python
mean=df["marks"].mean()
std=df["marks"].std(ddof=0)
l_l=mean-(3*std)
u_l=mean+(3*std)
print("mean",mean)
print("sd",std)
print("l_l",l_l)
print("u_l",u_l)
```

```python
df["marks_treated"]=df["marks"].clip(
    lower=l_l,
    upper=u_l
)
print("after cap")
print(df)
```
