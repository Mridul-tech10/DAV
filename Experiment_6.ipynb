```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
data={
    "studyhours":[1,2,3,4],
    "previousmarks":[2,3,5,4]
}
df=pd.DataFrame(data)
print("original dataset")
print(df)
scalar=StandardScaler()
standardized_data=scalar.fit_transform(df)
standardized_df=pd.DataFrame(
    standardized_data,
    columns=df.columns
)
print("\nstandarized dataset")
print(standardized_data.round(6))
covariance_matrix=np.cov(standardized_data,
                         rowvar=False)
print("\ncovariance matrix")
print(covariance_matrix.round(6))
```

```python
eigenvalues,eigenvectors=np.linalg.eig(
    covariance_matrix
)
eigenvalues=eigenvalues.real
eigenvectors=eigenvectors.real
print("eigenvalues")
print(eigenvalues.round(6))
print("\n eigenvector")
print(eigenvectors.round(6))
```

```python
index=np.argsort(eigenvalues)[::-1]
eigenvalues=eigenvalues[index]
eigenvectors=eigenvectors[:,index]
print("sorted eigenvalues")
print(eigenvalues.round(6))
print("\n sorted eigenvectors")
print(eigenvectors.round(6))
```

```python
explained_variance=(
    eigenvalues/eigenvalues.sum()
)
print("explained variance ratio")
print(explained_variance.round(6))
```

```python
cumulative_variance=np.cumsum(
    explained_variance
)
print("cumulated explained variance")
print(cumulative_variance.round(6))
```

```python
principal_components=np.dot(standardized_data,
                            eigenvectors)
pca_df=pd.DataFrame(
    principal_components,
    columns=["PC1","PC2"]
)
print("principal components")
print(pca_df.round(6))
```

```python
reduced_data=principal_components[:,0]
reduced_df=pd.DataFrame(
    reduced_data,
    columns=["PC1"]
)
print("reduced dataset")
print(reduced_df.round(6))
```
